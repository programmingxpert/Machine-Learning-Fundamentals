```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("Finance_Analytics_Dataset.csv")
```

```python
category_counts = df["Category"].value_counts()
print(category_counts)
```

```python
# Show the number of transactions in each category using count plot
sns.countplot(data = df, x="Category")
plt.xlabel("Category")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=45)
plt.show()
```

```python
# Show the number of transactions for each payment mode using a count plot
sns.countplot(data=df, x = "PaymentMode")
plt.title("Number of transactions by Payment mode")
plt.xlabel("Payment mode")
plt.ylabel("Number of transactions")
plt.show()
```

```python
# Show number of transactions by transaction type (Debit and Credit) using count plot
sns.countplot(data=df, x= "TransactionType")
plt.title("Number of transactions by Transaction type")
plt.show()
```

```python
# Debit vs credit amount comparison (bar plot)
sns.barplot(data = df, x= "TransactionType", y= "Amount")
plt.show()
```

```python
# Spending pattern by Customer Segment (Bar plot)
sns.barplot(data=df, x = "CustomerSegment", y= "Amount")
plt.show()
```

```python
# Payment mode usage distribution (pie chart)
paymentmode_counts = df['PaymentMode'].value_counts()
plt.pie(paymentmode_counts, labels= paymentmode_counts.index, autopct="%1.1f%%")
plt.show()
```

```python
# Boxplot of Transaction Amount by Category
sns.boxplot(data=df, x = "Amount", y = "Category")
plt.show()
```
